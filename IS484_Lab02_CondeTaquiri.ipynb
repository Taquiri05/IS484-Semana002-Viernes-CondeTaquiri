## Representación de un problema mediante estados

**Parte A:** Completar la ficha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| Estado | Descripción |
|---|---|
| **Estado inicial** | El repartidor se encuentra en el Almacén |
| **Estado objetivo** | El repartidor ha llegado a la Zona Norte |
| **Acciones posibles** | El repartidor se puede desplazar a cualquier punto que esté conectado directamente al punto en el que se encuentra |
| **Costo de cada acción** | El costo de cada acción (cada tramo recorrido) es el mismo para todos los desplazamientos |

       

```python
# Diccionario que representa el mapa de la ciudad como un grafo
# Cada clave es un punto (nodo), y su valor es la lista de puntos conectados directamente a el
mapa = {
    "Almacen": ["Centro", "Terminal"],       # desde Almacen se puede ir a Centro o Terminal
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)  # mostramos el grafo para verificar que esta bien definido
```

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    Explora el grafo nivel por nivel (por anillos).
    """
    cola = [[inicio]]   # la cola guarda CAMINOS completos, no solo nodos; empieza con el nodo inicial
    visitados = set()   # guardamos los nodos ya visitados para no repetir trabajo

    while cola:  # mientras haya caminos pendientes de explorar
        camino = cola.pop(0)       # saca el PRIMER camino de la cola (FIFO = el primero en entrar, sale primero)
        nodo_actual = camino[-1]   # el nodo actual es el ultimo de ese camino

        if nodo_actual == objetivo:   # si llegamos a la meta
            return camino              # devolvemos el camino completo que nos trajo hasta aqui

        if nodo_actual not in visitados:   # solo expandimos nodos que no hemos visitado antes
            visitados.add(nodo_actual)      # marcamos el nodo actual como visitado
            for vecino in grafo[nodo_actual]:      # recorremos todos los vecinos del nodo actual
                nuevo_camino = camino + [vecino]     # creamos un nuevo camino: el actual + el vecino nuevo
                cola.append(nuevo_camino)             # lo agregamos al FINAL de la cola, para explorarlo despues

    return None  # si la cola se vacia sin encontrar la meta, no hay camino posible

resultado = bfs(mapa, "Almacen", "ZonaNorte")   # ejecutamos BFS desde Almacen hasta ZonaNorte
print("Camino encontrado (BFS):", resultado)     # mostramos el camino encontrado
```

## Implementación de DFS

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    Explora un camino hasta el final antes de retroceder a probar otro.
    """
    pila = [[inicio]]   # la pila tambien guarda caminos completos; empieza con el nodo inicial
    visitados = set()

    while pila:  # mientras haya caminos pendientes
        camino = pila.pop()        # saca el ULTIMO camino agregado a la pila (LIFO = el ultimo en entrar, sale primero)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)   # se agrega al final, pero como es pila, sera el PROXIMO en salir

    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

```python
import numpy as np  # importamos numpy para trabajar con vectores y matrices

# Cada observacion es un vector con 2 caracteristicas: temperatura y humedad
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# Juntamos las 4 observaciones en una sola matriz (dataset)
# cada FILA es una observacion, cada COLUMNA es una caracteristica
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)  # (4, 2) = 4 observaciones, 2 caracteristicas

print("Primera observacion (fila 0):", dataset[0])       # accedemos a una fila completa
print("Todas las temperaturas (columna 0):", dataset[:, 0])  # accedemos a una columna completa (todas las filas, columna 0)
```

```python
# Producto escalar: multiplica elemento por elemento y suma el resultado
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)

# Distancia euclidiana calculada a mano, siguiendo la formula: raiz de la suma de las diferencias al cuadrado
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# La misma distancia, pero usando la funcion ya lista de NumPy (debe dar el mismo resultado)
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

```python
nueva_observacion = np.array([20.5, 48])  # una observacion nueva que queremos comparar contra el dataset

distancias = []  # aqui guardamos la distancia de la nueva observacion contra cada una del dataset
for i, obs in enumerate(dataset):              # recorremos cada observacion del dataset con su indice i
    d = np.linalg.norm(nueva_observacion - obs)  # calculamos la distancia euclidiana
    distancias.append(d)                          # la guardamos en la lista
    print(f"Distancia a la observacion {i}: {d}")

indice_mas_cercano = np.argmin(distancias)  # argmin devuelve el INDICE del valor mas pequeño de la lista
print("La observacion mas parecida es la numero", indice_mas_cercano)
```

## Reflexión final

La distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones porque toma en cuenta todas las características a la vez, calculando la separación real entre ambas: cuanto menor es esa distancia, más parecidas son.

Sin embargo, puede no ser la mejor opción cuando las características tienen escalas muy distintas (por ejemplo, soles vs. años). En ese caso, la variable con valores numéricamente más grandes domina el cálculo, distorsionando la comparación, aunque no sea la más relevante.
